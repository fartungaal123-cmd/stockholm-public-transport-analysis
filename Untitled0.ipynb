# Stockholm Public Transport Analysis

## 1. Project Overview

This project analyzes Stockholm public transport timetable data using Python and pandas.

The analysis focuses on SL (Storstockholms Lokaltrafik) and explores scheduled departures, differences between weekdays and weekends, and the distribution of scheduled trips across routes.

The goal is to practice data cleaning, data validation, data aggregation, and exploratory data analysis using real-world public transport data.

## Objectives

- Explore and validate GTFS timetable data.
- Analyze scheduled departures by hour of the day.
- Compare weekday and weekend departure patterns.
- Identify routes with a high average number of scheduled trips per date.
- Document data-quality issues and analytical limitations.

## Tools and Technologies

- Python
- pandas
- Google Colab
- Matplotlib
- GTFS static timetable data

## Important Note

This project analyzes scheduled timetable data. It does not measure passenger numbers, real-time delays, or whether scheduled trips actually operated.

```python

import pandas as pd

# Läs in tidtabellens hållplatstider
stop_times = pd.read_csv(
    "/content/stop_times.txt",
    dtype={"trip_id": str, "stop_id": str}
)

# Visa de första fem raderna
display(stop_times.head())

# Visa antal rader och kolumner
print("Antal rader:", len(stop_times))
print("Antal kolumner:", len(stop_times.columns))
```

## 2. Data Loading

The project uses static GTFS (General Transit Feed Specification) data from Trafiklab GTFS Sweden 3, published by Samtrafiken i Sverige AB.

The dataset contains timetable information, including routes, trips, stops, stop times, and service calendar information.

The files are loaded into pandas DataFrames for inspection, filtering, joining, and analysis.

The analysis focuses on SL services identified through the agency and route information in the dataset.

```python

# Läs in turer och linjer
trips = pd.read_csv(
    "/content/trips.txt",
    dtype={"trip_id": str, "route_id": str, "service_id": str}
)

routes = pd.read_csv(
    "/content/routes.txt",
    dtype={"route_id": str, "agency_id": str}
)

print("Antal rader i stop_times:", len(stop_times))
print("Antal unika turer i stop_times:", stop_times["trip_id"].nunique())
print("Antal rader i trips:", len(trips))
print("Antal unika linjer i routes:", routes["route_id"].nunique())

print("\nFörsta fem raderna i trips:")
display(trips.head())

print("\nFörsta fem raderna i routes:")
display(routes.head())
```

## 3. Data Cleaning and Validation

Before analyzing the timetable data, I checked the relationships between trips, routes, stop times, and service dates.

The validation focused on:

- Checking whether SL trips had associated stop-time records.
- Examining service calendar dates and exceptions.
- Comparing the number of scheduled trips with the number of trips that had identifiable starting departures.
- Investigating missing records to understand their impact on the analysis.

These checks help identify data-quality limitations and ensure that the analysis uses clearly defined records.

### Data Quality Finding

Some SL trips did not have associated stop-time records in the downloaded dataset. These trips were excluded from analyses that required a known departure time.

The difference between the total number of scheduled trips and the number of trips with identifiable starting departures was checked for each service date.

```python

# Hitta operatörsposten vi undersökte tidigare
sl_routes = routes[
    routes["agency_id"].astype(str) == "505000000000000001"
].copy()

print("Antal linjer kopplade till denna agency_id:", len(sl_routes))

print("\nExempel på linjer:")
display(
    sl_routes[
        ["route_id", "agency_id", "route_short_name", "route_type"]
    ].head(10)
)
```

```python

# Filtrera fram turer som hör till våra utvalda SL-linjer
sl_trips = trips[trips["route_id"].isin(sl_routes["route_id"])].copy()

print("Antal SL-turer i trips.txt:", len(sl_trips))
print("Antal unika SL-turer:", sl_trips["trip_id"].nunique())

# Kontrollera hur många av dessa turer som finns i stop_times.txt
stop_trip_ids = set(stop_times["trip_id"].astype(str))
sl_trips["trip_id"] = sl_trips["trip_id"].astype(str)

sl_trips_with_times = sl_trips[
    sl_trips["trip_id"].isin(stop_trip_ids)
].copy()

print("SL-turer som har hållplatstider:", len(sl_trips_with_times))
```

```python

# Behåll endast hållplatstider som tillhör våra SL-turer
sl_stop_times = stop_times[
    stop_times["trip_id"].astype(str).isin(
        sl_trips_with_times["trip_id"]
    )
].copy()

# Sortera hållplatserna i varje tur
sl_stop_times = sl_stop_times.sort_values(
    ["trip_id", "stop_sequence"]
)

# Hämta första hållplatsbesöket för varje tur
first_stops = sl_stop_times.drop_duplicates(
    subset="trip_id",
    keep="first"
).copy()

print("Antal SL-turer med hållplatstider:", len(first_stops))
print("\nExempel på första hållplatsbesök:")
display(
    first_stops[
        ["trip_id", "departure_time", "stop_id", "stop_sequence"]
    ].head(10)
)
```

```python

# Gör om avgångstiderna till text och hämta timmen
first_stops["departure_hour"] = (
    first_stops["departure_time"]
    .astype(str)
    .str.split(":")
    .str[0]
)

# Räkna antal turstarter per timme
departures_per_hour = (
    first_stops.groupby("departure_hour")
    .size()
    .reset_index(name="planned_trip_starts")
)

# Sortera timmarna i tidsordning
departures_per_hour = departures_per_hour.sort_values(
    "departure_hour"
)

display(departures_per_hour)
```

```python

# Gör om avgångstiden till en vanlig timme på dygnet
first_stops["hour_of_day"] = (
    first_stops["departure_time"]
    .astype(str)
    .str.split(":")
    .str[0]
    .astype(int) % 24
)

# Räkna planerade turstarter per vanlig klocktimme
departures_by_clock_hour = (
    first_stops.groupby("hour_of_day")
    .size()
    .reindex(range(24), fill_value=0)
    .reset_index(name="planned_trip_starts")
)

display(departures_by_clock_hour)
```

## 4. Analysis of Scheduled Departures

This section examines how scheduled departures from SL services are distributed across the hours of the day.

The departure time of each trip was extracted from the GTFS stop-time data. Since GTFS times can exceed 24:00, the hour was converted to a 24-hour clock using modulo 24.

Service dates were classified as weekdays or weekends. The analysis then calculated the average number of scheduled trip starts for each hour and day type across the analyzed dates.

### Key Findings

- **Weekdays:** The highest average number of scheduled trip starts occurred at 07:00, with a second peak at 16:00.
- **Weekends:** The highest average occurred at 14:00, followed closely by 15:00.
- **Overall pattern:** Weekday departures show clearer morning and afternoon peaks, while weekend departures are more evenly distributed throughout the day.

These results describe scheduled trip starts, not passenger volumes or actual service performance.

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(
    departures_by_clock_hour["hour_of_day"],
    departures_by_clock_hour["planned_trip_starts"]
)

plt.title("Planerade turstarter per timme – SL")
plt.xlabel("Timme på dygnet")
plt.ylabel("Antal planerade turstarter")
plt.xticks(range(24))
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. Route Analysis

This section compares SL routes based on their average number of scheduled trips per service date.

The trip data was combined with route information and service calendar dates. To ensure a fair comparison across the analyzed period, all routes were evaluated across the same 55 dates, with dates without scheduled trips counted as zero.

### Key Findings

The Red Line route 14 had the highest average, with approximately 371 scheduled trips per date.

Other routes with high average scheduled trip counts included Red Line route 13, Tvärbanan route 30, and Green Line routes 18, 17, and 19.

The results describe the number of scheduled trips assigned to each route in the downloaded timetable data. They do not indicate passenger demand, route capacity, or actual service performance.

```python
# Kontrollera hur många resor som hör till varje service_id
service_trip_counts = (
    sl_trips
    .groupby("service_id")["trip_id"]
    .nunique()
    .reset_index(name="number_of_trips")
)

# Lägg till antalet datum per service_id
service_trip_counts = service_trip_counts.merge(
    dates_per_service,
    on="service_id",
    how="left"
)

# Beräkna hur många resstarter varje service_id bidrar med
# över samtliga datum i den nuvarande kalenderkopplingen
service_trip_counts["trip_date_combinations"] = (
    service_trip_counts["number_of_trips"]
    * service_trip_counts["number_of_dates"]
)

display(
    service_trip_counts.sort_values(
        "trip_date_combinations",
        ascending=False
    ).head(10)
)

print(
    "Summa kombinationer:",
    service_trip_counts["trip_date_combinations"].sum()
)

print(
    "Faktiskt antal rader i kopplingen:",
    len(sl_trip_starts_dates)
)
```

```python
# Räkna faktiska kombinationer av resa och datum per service_id
actual_combinations = (
    sl_trip_starts_dates
    .groupby("service_id")
    .size()
    .reset_index(name="actual_combinations")
)

# Jämför med den teoretiska beräkningen
comparison = service_trip_counts.merge(
    actual_combinations,
    on="service_id",
    how="left"
)

comparison["actual_combinations"] = (
    comparison["actual_combinations"].fillna(0).astype(int)
)

comparison["difference"] = (
    comparison["trip_date_combinations"]
    - comparison["actual_combinations"]
)

# Visa service-ID:n där det finns en skillnad
display(
    comparison[comparison["difference"] != 0]
    .sort_values("difference", ascending=False)
    .head(15)
)

print(
    "Summa skillnader:",
    comparison["difference"].sum()
)
```

```python
# Hitta service-ID:n med resor men inga kopplade avgångar
missing_starts = comparison[
    comparison["actual_combinations"] == 0
].copy()

display(
    missing_starts[
        [
            "service_id",
            "number_of_trips",
            "number_of_dates",
            "trip_date_combinations",
            "actual_combinations"
        ]
    ].sort_values("number_of_trips", ascending=False)
)

print(
    "Antal service-ID:n utan kopplade avgångar:",
    len(missing_starts)
)
```

```python
# Hämta service-ID:n som saknar kopplade avgångar
missing_service_ids = set(
    comparison.loc[
        comparison["actual_combinations"] == 0,
        "service_id"
    ].astype(str)
)

# Visa resorna för dessa service-ID:n
missing_trips = sl_trips[
    sl_trips["service_id"].astype(str).isin(missing_service_ids)
].copy()

# Kontrollera om deras trip_id finns i stop_times
stop_time_trip_ids = set(stop_times["trip_id"].astype(str))

missing_trips["has_stop_times"] = (
    missing_trips["trip_id"].astype(str).isin(stop_time_trip_ids)
)

# Sammanfatta per service-ID
diagnosis = (
    missing_trips.groupby("service_id")
    .agg(
        number_of_trips=("trip_id", "nunique"),
        trips_with_stop_times=("has_stop_times", "sum")
    )
    .reset_index()
)

diagnosis["trips_without_stop_times"] = (
    diagnosis["number_of_trips"]
    - diagnosis["trips_with_stop_times"]
)

display(diagnosis.sort_values("number_of_trips", ascending=False))

print("Totalt antal resor:", diagnosis["number_of_trips"].sum())
print("Resor med stop_times:", diagnosis["trips_with_stop_times"].sum())
print("Resor utan stop_times:", diagnosis["trips_without_stop_times"].sum())
```

```python
# Alla SL-resor på de berörda linjerna
affected_routes = missing_by_route["route_id"].astype(str)

trips_on_affected_routes = sl_trips[
    sl_trips["route_id"].astype(str).isin(affected_routes)
].copy()

# Kontrollera om varje resa har minst en rad i stop_times
stop_time_trip_ids = set(stop_times["trip_id"].astype(str))

trips_on_affected_routes["has_stop_times"] = (
    trips_on_affected_routes["trip_id"]
    .astype(str)
    .isin(stop_time_trip_ids)
)

# Sammanfatta per linje
route_check = (
    trips_on_affected_routes
    .groupby("route_id")
    .agg(
        total_trips=("trip_id", "nunique"),
        trips_with_stop_times=("has_stop_times", "sum")
    )
    .reset_index()
)

route_check["trips_without_stop_times"] = (
    route_check["total_trips"]
    - route_check["trips_with_stop_times"]
)

route_check = route_check.merge(
    routes[["route_id", "route_short_name", "route_long_name"]],
    on="route_id",
    how="left"
)

display(
    route_check[
        [
            "route_short_name",
            "route_long_name",
            "total_trips",
            "trips_with_stop_times",
            "trips_without_stop_times"
        ]
    ].sort_values("trips_without_stop_times", ascending=False)
)
```

```python
# Alla SL-linjer
route_list = sl_routes[
    ["route_id", "route_short_name", "route_long_name"]
].drop_duplicates()

# Alla 55 datum i kalenderunderlaget
all_dates = pd.DataFrame({
    "date": sorted(sl_calendar_dates["date"].unique())
})

# Skapa alla kombinationer av linje och datum
route_list["key"] = 1
all_dates["key"] = 1

route_date_grid = (
    route_list
    .merge(all_dates, on="key")
    .drop(columns="key")
)

# Koppla in antalet planerade resor per linje och datum
route_date_complete = route_date_grid.merge(
    daily_route_counts[
        ["date", "route_id", "scheduled_trips"]
    ],
    on=["date", "route_id"],
    how="left"
)

# Datum utan registrerade resor får värdet 0
route_date_complete["scheduled_trips"] = (
    route_date_complete["scheduled_trips"].fillna(0)
)

# Beräkna genomsnittet över samtliga datum
route_average_all_dates = (
    route_date_complete
    .groupby(
        ["route_id", "route_short_name", "route_long_name"]
    )
    .agg(
        avg_scheduled_trips_per_date=("scheduled_trips", "mean"),
        antal_datum_med_resor=(
            "scheduled_trips",
            lambda x: (x > 0).sum()
        )
    )
    .reset_index()
    .sort_values("avg_scheduled_trips_per_date", ascending=False)
)

display(route_average_all_dates.head(10))
```

## 6. Key Findings and Limitations

### Key Findings

- Scheduled trip starts on weekdays peaked at 07:00, with a second peak at 16:00.
- Weekend departures were more evenly distributed, with the highest average at 14:00.
- SL route 14 on the Red Line had the highest average number of scheduled trips per date among the analyzed routes.
- Data validation identified SL trips without associated stop-time records, highlighting the importance of checking data completeness before analysis.

### Limitations

- The dataset contains static timetable information, not real-time operational data.
- Scheduled trips do not necessarily represent trips that actually operated.
- The dataset does not contain passenger counts, so passenger demand cannot be measured.
- Trips without stop-time records could not be included in analyses requiring identifiable departure times.
- The findings reflect the dates and timetable version included in the downloaded GTFS feed.
